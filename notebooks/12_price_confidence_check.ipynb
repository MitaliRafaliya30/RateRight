```python
import duckdb

cols = duckdb.sql("""
    DESCRIBE SELECT * FROM read_parquet('../data/gold/nyc/listings_prepared.parquet')
""").df()
print(cols["column_name"].tolist())
```

```python
diagnosis = duckdb.sql("""
    SELECT
        base_price IS NOT NULL AS has_price,
        price_confidence,
        COUNT(*) AS listings
    FROM read_parquet('../data/gold/nyc/listings_prepared.parquet')
    GROUP BY has_price, price_confidence
    ORDER BY has_price
""").df()
print(diagnosis)
```

```python
detail = duckdb.sql("""
    SELECT room_type, stay_type, price_confidence, COUNT(*) AS listings
    FROM read_parquet('../data/gold/nyc/listings_prepared.parquet')
    WHERE base_price IS NOT NULL
    GROUP BY room_type, stay_type, price_confidence
    ORDER BY listings
""").df()
print(detail)
```
